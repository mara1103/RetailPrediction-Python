```python
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
```

```python
# pip install tensorflow
```

```python
df_lstm = df.set_index("DATA").asfreq("D")
```

```python
df_lstm
```

```python
split_ratio = None  # using last 14 days as test
n = len(df_lstm)
split_idx = n - 14

split_idx
```

```python
train = df_lstm.iloc[:split_idx]
test  = df_lstm.iloc[split_idx:]
```

```python
# Scaling - doar pe TRAIN TEST
```

```python
scaler = MinMaxScaler()
train_scaled = scaler.fit_transform(train[["CANTITATE"]])
full_scaled = scaler.transform(df_lstm[["CANTITATE"]])
```

```python
def make_sequences(data_scaled, lookback=14):
    X, y = [], []
    for i in range(lookback, len(data_scaled)):
        X.append(data_scaled[i-lookback:i, 0])  # (lookback,)
        y.append(data_scaled[i, 0])             # scalar
    X = np.array(X)
    y = np.array(y)
    return X[..., np.newaxis], np.array(y)      # X shape: (samples, lookback, 1)

lookback = 21
X_all, y_all = make_sequences(full_scaled, lookback)
target_idx = np.arange(lookback, n)
train_mask = target_idx < split_idx
X_train, y_train = X_all[train_mask], y_all[train_mask]
X_test, y_test   = X_all[~train_mask], y_all[~train_mask]

X_train.shape, X_test.shape
```

```python
tf.random.set_seed(42)

model = Sequential([
    LSTM(64, input_shape=(lookback, 1), return_sequences=False),
    Dropout(0.2),
    Dense(32, activation="relu"),
    Dense(1)
])

model.compile(optimizer="adam", loss="mse")

es = EarlyStopping(monitor="val_loss", patience=12, restore_best_weights=True)

# Learning - 

history = model.fit(
    X_train, y_train,
    validation_split=0.2,
    epochs=100,
    shuffle=False,
    batch_size=32,
    callbacks=[es],
    verbose=1
)
```

```python
pred_scaled = model.predict(X_test).reshape(-1, 1)

pred = scaler.inverse_transform(pred_scaled).flatten()
y_true = scaler.inverse_transform(y_test.reshape(-1, 1)).flatten()
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

mae = mean_absolute_error(y_true, pred)

mse = mean_squared_error(y_true, pred)
rmse = np.sqrt(mse)

mape = np.mean(np.abs((y_true - pred) / np.maximum(y_true, 1))) * 100


mae = float(mae)
rmse = float(rmse)
mape = float(mape)


mae, rmse, mape

```

### Interpretarea performanței modelului LSTM (lookback = 21)

Modelul LSTM multivariant, utilizând o fereastră temporală de 21 de zile, a fost evaluat pe setul de testare folosind metrici specifice problemelor de regresie și prognoză a seriilor temporale.

Eroarea medie absolută (MAE) obținută este de 5,25 unități, ceea ce indică faptul că, în medie, predicțiile modelului diferă de valorile reale ale cantității vândute cu aproximativ cinci unități pe zi. Având în vedere nivelul și variabilitatea cererii din datele analizate, această valoare poate fi considerată satisfăcătoare și adecvată pentru scopuri de planificare a stocurilor.

Eroarea pătratică medie (RMSE) are valoarea de 8,07 unități, fiind mai mare decât MAE, ceea ce sugerează existența unor erori mai mari în cazul variațiilor extreme ale cererii. Acest comportament este tipic pentru datele din retail, caracterizate prin fluctuații ocazionale și vârfuri izolate, pe care modelul LSTM tinde să le atenueze.

Coeficientul de determinare (R²) este de 0,54, indicând faptul că aproximativ 54% din variația cantității vândute este explicată de model. Pentru o serie temporală reală, cu zgomot ridicat și influențe externe dificil de modelat, această valoare reflectă o capacitate bună a modelului de a surprinde tendința generală a cererii.

Coeficientul de corelație Pearson (r) are valoarea de 0,73, ceea ce arată o corelație pozitivă puternică între valorile reale și cele estimate. Aceasta confirmă faptul că modelul urmărește corect direcția și dinamica generală a cererii, chiar dacă nu reproduce cu exactitate amplitudinea tuturor variațiilor zilnice.

În ceea ce privește MAPE, valoarea ridicată obținută (83,7%) este explicată de prezența valorilor zero sau foarte mici în seria analizată, care conduc la distorsionarea acestui indicator. Din acest motiv, MAPE nu a fost utilizată ca metrică principală în evaluarea performanței modelului.

### Concluzie 

În ansamblu, modelul LSTM multivariant cu lookback de 21 de zile prezintă o performanță adecvată pentru estimarea nivelului mediu al cererii și pentru identificarea tendințelor generale, fiind util în contextul planificării stocurilor. Deși modelul nu surprinde în mod exact variațiile extreme izolate, comportamentul său conservator contribuie la obținerea unor predicții stabile și robuste, relevante pentru decizii operaționale.
